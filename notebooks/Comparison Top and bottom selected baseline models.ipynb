## Export Top Dataset 
Run the LR/RF/XGB definition cells first, then run the next cell to save the filtered dataset.
Output file: `data/top_dataset.csv`

```python
from pathlib import Path
import pandas as pd

# Use exactly the provided top sets
lr = {
    "clean2", "analcatdata_authorship", "clean1", "texture", "prnn_crabs", "penguins",
    "mfeat_factors", "dermatology", "wine_recognition", "twonorm", "optdigits",
    "mfeat_pixel", "soybean", "chess", "kr_vs_kp", "mofn_3_7_10", "mfeat_karhunen",
    "dna", "agaricus_lepiota", "mushroom", "segmentation", "pendigits", "spambase",
    "labor", "tokyo1", "iris", "waveform_40", "waveform_21", "ecoli", "mfeat_zernike"
}
rf = {
    "clean1", "clean2", "mux6", "mushroom", "agaricus_lepiota", "irish", "xd6", "collins",
    "corral", "ann_thyroid", "penguins", "analcatdata_creditscore", "monk3", "threeOf9",
    "soybean", "pendigits", "segmentation", "optdigits", "analcatdata_authorship", "dermatology",
    "wine_recognition", "texture", "mfeat_pixel", "vowel", "kr_vs_kp", "chess", "iris",
    "mfeat_karhunen", "prnn_crabs", "splice"
}
xgb = {
    "clean1", "irish", "mushroom", "clean2", "analcatdata_creditscore", "monk3",
    "agaricus_lepiota", "analcatdata_lawsuit", "penguins", "mfeat_karhunen", "mfeat_pixel",
    "confidence", "prnn_crabs", "vowel", "optdigits", "analcatdata_bankruptcy", "mfeat_factors",
    "analcatdata_authorship", "mfeat_zernike", "mfeat_fourier", "tokyo1", "waveform_21",
    "prnn_synth", "splice", "molecular_biology_promoters", "spambase", "cars", "waveform_40",
    "labor", "pendigits"
}

top_dataset = sorted(lr | rf | xgb)

# Read full manifest
manifest = pd.read_csv("dataset_manifest.csv")
manifest["dataset_base"] = manifest["dataset"].str.replace(".tsv.gz", "", regex=False)

# Keep only metadata/feature columns
meta_feature_cols = [
    "dataset", "target_column", "rows", "features", "classes", "class_imbalance",
    "missing_values_count", "missing_values_ratio", "numeric_features",
    "categorical_features", "feature_type"
]

top_dataset_df = manifest[manifest["dataset_base"].isin(top_dataset)][meta_feature_cols].copy()

# Save
out_path = Path("data") / "top_dataset.csv"
out_path.parent.mkdir(exist_ok=True)
top_dataset_df.to_csv(out_path, index=False)

print(f"Saved: {out_path}")
print(f"Rows: {len(top_dataset_df)}")
print(top_dataset_df.head())
```
